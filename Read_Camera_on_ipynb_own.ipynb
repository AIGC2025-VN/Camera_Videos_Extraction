```python
# declared libraries
import cv2
import IPython.display as display
from PIL import Image
import time
```

```python
# set up to read frames, using own camera on device running this code
cap = cv2.VideoCapture(0)

try:
    while True:
        ret, frame = cap.read()
        if not ret:
            break

        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)  # Chuyển ảnh sang RGB
        img = Image.fromarray(frame)  # Chuyển đổi thành ảnh PIL

        display.clear_output(wait=True)  # Xóa frame cũ
        display.display(img)  # Hiển thị frame mới

        time.sleep(0.05)  # Giữ tốc độ hiển thị

except KeyboardInterrupt:
    print("Dừng quay video.")
```
